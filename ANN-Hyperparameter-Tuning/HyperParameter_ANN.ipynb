```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv("/content/diabetes.csv")
```

```python
df.sample(5)
```

```python
df.corr()['Outcome']
```

```python
X = df.iloc[:,:-1].values
y = df.iloc[:,-1].values
```

```python
X
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X = sc.fit_transform(X)
```

```python
X
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
import tensorflow
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(32, activation='relu', input_dim=8))
model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer='Adam',loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=10, batch_size=32,validation_data=(X_test, y_test))
```

```python
import kerastuner as kt
```

```python
#building model from scratch
def build_model(hp):
  model = Sequential()
  model.add(Dense(32, activation='relu', input_dim=8))
  model.add(Dense(1, activation='sigmoid'))

  optimizers = hp.Choice('optimizer',values=['adam','sgd','rmsprop','adadelta'])

  model.compile(optimizer=optimizers,loss='binary_crossentropy', metrics=['accuracy'])

  return model
```

```python
tuner = kt.RandomSearch(build_model,objective='val_accuracy',max_trials=5)
```

```python
tuner.search(X_train, y_train, epochs=5, validation_data=(X_test, y_test))
```

```python
tuner.results_summary()
```

```python
tuner.get_best_hyperparameters()[0].values
```

```python
model = tuner.get_best_models(num_models=1)[0]
```

```python
model.summary()
```

```python
model.fit(X_train, y_train,batch_size=32, epochs=100, validation_data=(X_test, y_test))
```

```python
def build_model(hp):
  model = Sequential()
  units = hp.Int('units',min_value=8,max_value=124,step=8)
  model.add(Dense(units=units, activation='relu', input_dim=8))
  model.add(Dense(1,activation='sigmoid'))

  model.compile(optimizer='adam',loss='binary_crossentropy', metrics=['accuracy'])

  return model
```

```python
tuner = kt.RandomSearch(build_model,objective='val_accuracy',max_trials=5,directory='mkdir')
```

```python
tuner.search(X_train, y_train, epochs=5, validation_data=(X_test, y_test))
```

```python
def build_model(hp):
  model = Sequential()

  model.add(Dense(32, activation='relu', input_dim=8))

  for i in range(hp.Int('num_layers',min_value=1,max_value=10)):
    model.add(Dense(32, activation='relu'))

  model.add(Dense(1,activation='sigmoid'))
  model.compile(optimizer='adam',loss='binary_crossentropy', metrics=['accuracy'])

  return model
```

```python
tuner = kt.RandomSearch(build_model,objective='val_accuracy',max_trials=5,directory='mkdir',project_name='layers')
```

```python
tuner.search(X_train, y_train, epochs=5, validation_data=(X_test, y_test))
```

```python
model = tuner.get_best_models(num_models=1)[0]
```
